```python
# Почти-константы (топ-категория >95%), категориальные признаки 


cat_summary = summary.loc[cat_cols].copy()

# Делаем новый столбец "число непустых значений"

cat_summary["count"] = len(train) - cat_summary["n_missing"]

# находим долю топ-значений среди непустых 
cat_summary["top_share"] = (cat_summary["top_freq"] / cat_summary["count"]).round(3)

# сортируем топ непустых 

cat_summary= cat_summary.sort_values("top_share", ascending = False)
cat_summary[["n_unique", "top", "top_freq", "top_share"]]


```

### Вывод по почти-константам

**Пороги:**
- `top_share > 0.95` — почти-константа, кандидат на удаление.
- `top_share` 0.9–0.95 — пограничные, смотреть по смыслу.
- `top_share < 0.9` — полезные признаки.

**Колонки с `top_share > 0.95` (8 штук):**
| Колонка | n_unique | top | top_share |
|---|---|---|---|
| Utilities | 2 | AllPub | 1.00 |
| Street | 2 | Pave | 1.00 |
| Condition2 | 8 | Norm | 0.99 |
| RoofMatl | 8 | CompShg | 0.98 |
| Heating | 6 | GasA | 0.98 |
| GarageCond | 5 | TA | 0.96 |

**Пример `Utilities`:**
- 1459 из 1460 домов имеют `AllPub`, 1 — `NoSeWa`.
- Группа `NoSeWa` = 1 дом. Статистически бессмысленно: средняя по одному дому — это цена этого дома.
- Признак не различает дома → кандидат на удаление.

```python
#Проверь числовые на почти-константность

num_summary = summary.loc[pure_num_cols].copy()
num_summary["range"] = num_summary["max"] - num_summary["min"] #диапазон значений, если  0, то все значения близки
num_summary["cv"]= (train[pure_num_cols].std() / train[pure_num_cols].mean()).round(3) #коэффициент вариации. Мера относительного разброса
num_summary["zero_share"] = (train[pure_num_cols] == 0).mean().round(3) #  Доля нулей (для счётных признаков)

num_summary.sort_values("cv").head(10) # Сверху — самые «стабильные» (потенциальные почти-константы).

#Что такое CV:

# CV = 0 → признак константа (std = 0).
# CV маленький (< 0.1) → почти константа.
# CV средний (0.1–1) → нормальный разброс.
# CV большой (> 1) → сильно разбросанный.


# zero_share > 0.95 → признак почти всегда 0 → мусор.
# zero_share 0.5–0.95 → признак различает «есть/нет» → может быть полезен.
```
